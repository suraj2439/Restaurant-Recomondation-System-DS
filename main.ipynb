# Examination of data distribution

```python
df = pd.read_csv("Assgn1-111903086_111903106_111903109.csv",encoding='cp1252')
```

```python
avgCost = {}
for i in range (0, len(df)):
    print(df["Average Cost"][i])
    val = df["Average Cost"][i].split(" for ")
    val[0] = val[0].replace(",", "")
    val[1] = val[1].rstrip()
    cost = int(val[0])
    #print(val[1])
    if("two" in val[1]):
        cost = int(val[0])/2
    avgCost[i] = cost
    #df["Average Cost"].replace(to_replace = i, value = 100)
df["Average Cost"] = df.index.map(avgCost)
plt.hist(df["Average Cost"], bins=50,width=100)
```

```python
df.replace("-", np.nan, inplace=True)
df.replace("New", np.nan, inplace=True)
df.replace('0', np.nan, inplace=True)

df.isnull().sum()
```

```python
pMode = {}

for i in range (0, len(df)):
    #print(df["Average Cost"][i])
    val = df["Payment Modes"][i]

    val1 = str(val)
    res = ""
    if(val1 == ""):
        res += 'Cash'
    if('Cash' in val1):
        res+='Cash'
    if('Card' in val1):
        res+='Card'
    if('Coupen' in val1):
        res+='Coupen'
    if('Digital' in val1):
        res+='Digital'
    
    pMode[i] = res
    #df["Payment Modes"].replace(val, res)

df["Payment Modes"] = df.index.map(pMode)
#df["Average Cost"] = df.index.map(avgCost)
#plt.hist(df["Average Cost"], bins=50,width=100)
```

```python
peopleSayCat = ["ambience", "service", "food", "budget", "delivery"]
ambience = ["ambience", "environment", "place", "feel", "seating", "atmosphere", "music", "venue","decor"]
budget = ["budget", "portion", "economical", "price", "worth", "money"]
food = ["variety", "taste", "menu", "food", "dessert","appetizer"]
service = ["service", "staff", "behaviour"]
delivery = ["delivery", "packaging"]

positive1 = ["good", "nice", "average", "class", "extensive", "plenty", "elaborat", "proper", "quick", "love", "inexpensive", "cheap" "filling", "casual", "cute", "tamper proof",
             "polite", "comfort", "great", "prompt", "relax", "fresh", "healthy", "reasonable","fair","contactless","clean"]
positive2 = ["yummy", "affordable", "cozy", "fast", "value", "timely", "delicious", "tasty", "hygienic", "decent", "amazing", "wonderful", "excellent", 
             "outstanding", "best", "fantastic","mouth watering","awesome","beautiful"]


def getPoints(key, lst, points):
    for ele in lst:
        if ele in key:
            if points == 0:
                return 5
            else:
                return points
    return 0

ambiencePoints = []
budgetPoints = []
foodPoints = []
servicePoints = []
deliveryPoints = []

for j in range (0, len(df)):
    ambienceRating, budgetRating, foodRating, serviceRating, deliveryRating = 0,0,0,0,0
    arr = df["People Say"][j]
    arr = arr.split(",")
    for i in range (len(arr)):
        arr[i] = arr[i].lower().strip()
        
    for remark in arr:
        points = 0
        for word in positive1:
            if word in remark:
                points = 7
                break
        for word in positive2:
            if word in remark:
                points = 10
                break
                
        ambienceRating += getPoints(remark, ambience, points)
        budgetRating += getPoints(remark, budget, points)
        foodRating += getPoints(remark, food, points)
        serviceRating += getPoints(remark, service, points)
        deliveryRating += getPoints(remark, delivery, points)
        
    #print(ambienceRating, budgetRating, foodRating, serviceRating, deliveryRating)
            
#     print(df["Average Cost"][i])
#     val = df["Average Cost"][i].split(" for ")
#     val[0] = val[0].replace(",", "")
#     val[1] = val[1].rstrip()
#     cost = int(val[0])
#     #print(val[1])
#     if("two" in val[1]):
#         cost = int(val[0])/2

    ambiencePoints.append(ambienceRating)
    budgetPoints.append(budgetRating)
    foodPoints.append(foodRating)
    servicePoints.append(serviceRating)
    deliveryPoints.append(deliveryRating)
    #df["Average Cost"].replace(to_replace = i, value = 100)

#print(tmp)
df["Ambience Rating"] = pd.Series(ambiencePoints)
df["Budget Rating"] = pd.Series(budgetPoints)
df["Food Rating"] = pd.Series(foodPoints)
df["Service Rating"] = pd.Series(servicePoints)
df["Delivery Rating"] = pd.Series(deliveryPoints)
print("done")
df
```
